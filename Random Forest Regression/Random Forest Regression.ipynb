```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy for sqrt

# Ensure df is defined - loading the specified CSV
# This assumes the user wants to proceed with the Video Games Sales data
df = pd.read_csv('/content/Video_Games_Sales_as_at_22_Dec_2016.csv')

# --- Data Preprocessing for Video Games Sales Dataset ---

# Make a copy to avoid modifying the original dataframe
df_processed = df.copy()

# Drop irrelevant columns and columns that sum up to Global_Sales
df_processed.drop(columns=['Name', 'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales'], errors='ignore', inplace=True)

# Handle missing values
# Drop rows where the target variable 'Global_Sales' is NaN
df_processed.dropna(subset=['Global_Sales'], inplace=True)

# Fill missing categorical features with 'Unknown'
for col in ['Publisher', 'Developer', 'Rating', 'Genre']:
    if col in df_processed.columns:
        # Reassign to avoid FutureWarning with inplace=True on a copy
        df_processed[col] = df_processed[col].fillna('Unknown')

# For Year_of_Release, fill with mode (most frequent year) and convert to int
if 'Year_of_Release' in df_processed.columns:
    # Reassign to avoid FutureWarning with inplace=True on a copy
    df_processed['Year_of_Release'] = df_processed['Year_of_Release'].fillna(df_processed['Year_of_Release'].mode()[0])
    df_processed['Year_of_Release'] = df_processed['Year_of_Release'].astype(int)

# Identify features (X) and target (y)
y = df_processed['Global_Sales']
X = df_processed.drop('Global_Sales', axis=1)

# One-hot encode categorical features
categorical_cols = X.select_dtypes(include='object').columns
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# --- Train-Test Split ---
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# --- Random Forest Regression Model ---
# Initialize the Random Forest Regressor
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)

# Train the model
rf_regressor.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_regressor.predict(X_test)

# --- Evaluation ---
mse = mean_squared_error(y_test, y_pred)
# Calculate RMSE by taking the square root of MSE
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Random Forest Regression Performance:")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-squared (R2): {r2:.4f}")

# --- Visualizations ---

plt.figure(figsize=(14, 6))

# 1. Actual vs. Predicted Values Scatter Plot
plt.subplot(1, 2, 1)
sns.scatterplot(x=y_test, y=y_pred, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2) # Line for perfect prediction
plt.title('Actual vs. Predicted Global Sales')
plt.xlabel('Actual Global Sales')
plt.ylabel('Predicted Global Sales')
plt.grid(True)

# 2. Feature Importances
plt.subplot(1, 2, 2)
feature_importances = pd.Series(rf_regressor.feature_importances_, index=X_encoded.columns)
top_n = 20 # Display top 20 features
feature_importances.nlargest(top_n).plot(kind='barh')
plt.title(f'Top {top_n} Feature Importances for Global Sales')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()

print("Note: The model predicts 'Global_Sales' based on other game attributes.")
```
