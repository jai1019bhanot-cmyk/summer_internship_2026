```python

# sentence-level and word-level tokenization meaningfully.

corpus = """The researchers are studying machine learning algorithms.
They have been writing papers and programming new models.
The history of AI goes back to the 1950s, and the field has finally matured."""

# Single sentence used for stemming & lemmatization comparison
sentence = "The researchers are studying, writing papers, programming models, and finalizing their history."

# Word list extracted from the sentence (same style as your course notebooks)
words = ["studying", "studies", "studied",
         "writing", "writes", "written",
         "programming", "programs", "programmed",
         "finalizing", "finalized", "finally",
         "history", "historically", "historical"]

print("Corpus:")
print(corpus)
print("\nWord list:", words)
```

```python

import nltk
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(corpus)

print("=== sent_tokenize: Paragraph → Sentences ===")
print(f"Total sentences found: {len(sentences)}\n")
for i, sent in enumerate(sentences, 1):
    print(f"Sentence {i}: {sent}")
```

```python

from nltk.tokenize import word_tokenize

word_tokens = word_tokenize(sentence)

print("=== word_tokenize: Sentence → Words ===")
print(f"Tokens: {word_tokens}")
print(f"\nTotal tokens: {len(word_tokens)}")
print("Note: punctuation (comma, full stop) appears as separate tokens.")
```

```python

from nltk.tokenize import wordpunct_tokenize

wp_tokens = wordpunct_tokenize(sentence)

print("=== wordpunct_tokenize ===")
print(f"Tokens: {wp_tokens}")
print(f"\nTotal tokens: {len(wp_tokens)}")
print("Note: Very similar to word_tokenize for standard English text.")
```
