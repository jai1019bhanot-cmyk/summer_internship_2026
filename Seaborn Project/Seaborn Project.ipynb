```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/content/healthcare.csv")

# Scatter plot: Age vs BMI, colored by Gender
sns.scatterplot(x="Age", y="BMI", hue="Gender", data=df)
plt.title("Scatter Plot: Age vs BMI")
plt.show()
```

```python
# Regression plot: Age vs Systolic Blood Pressure
sns.regplot(x="Age", y="BloodPressure_Systolic", data=df, scatter_kws={"alpha":0.5})
plt.title("Regression Plot: Age vs Systolic BP")
plt.show()
```

```python
# Count plot: frequency of each disease
sns.countplot(x="Disease", data=df, order=df['Disease'].value_counts().index)
plt.xticks(rotation=90)
plt.title("Count Plot: Disease Frequency")
plt.show()
```

```python
# Bar plot: average hospital stay days per disease
avg_stay = df.groupby("Disease")["HospitalStayDays"].mean().reset_index()
sns.barplot(x="Disease", y="HospitalStayDays", data=avg_stay, order=avg_stay.sort_values("HospitalStayDays")["Disease"])
plt.xticks(rotation=90)
plt.title("Bar Plot: Avg Hospital Stay by Disease")
plt.show()
```

```python
# Heatmap: correlation between numeric health features
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Healthcare Features")
plt.show()
```
