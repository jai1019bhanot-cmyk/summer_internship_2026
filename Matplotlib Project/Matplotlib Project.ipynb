```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/Fish.csv")

# Line plot: Length vs Weight trend
plt.plot(df['Length3'], df['Weight'], marker='o', linestyle='-', color='blue')

plt.xlabel("Length3 (cm)")        # X-axis label
plt.ylabel("Weight (grams)")      # Y-axis label
plt.title("Fish Length vs Weight (Line Plot)")  # Title
plt.grid(True)                    # Add grid for clarity
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("/content/Fish.csv")

# Area plot: stacking Length1, Length2, Length3 for first 50 rows
plt.stackplot(
    df.index[:50],
    df['Length1'].iloc[:50],
    df['Length2'].iloc[:50],
    df['Length3'].iloc[:50],
    labels=['Length1','Length2','Length3']
)

plt.xlabel("Index (first 50 samples)")
plt.ylabel("Length (cm)")
plt.title("Area Plot of Fish Lengths")
plt.legend(loc='upper left')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/Fish.csv")

# Histogram shows distribution of fish weights
plt.hist(df['Weight'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel("Weight (grams)")
plt.ylabel("Frequency")
plt.title("Distribution of Fish Weights")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/Fish.csv")

plt.figure(figsize=(14,10))

ax1=plt.subplot(331)
ax2=plt.subplot(332)
ax3=plt.subplot(333)
ax4=plt.subplot(334)
ax5=plt.subplot(335)
ax6=plt.subplot(336)
ax7=plt.subplot(337)
ax8=plt.subplot(338)
ax9=plt.subplot(339)

# 1. Line Plot → Length vs Weight trend
ax1.plot(df['Length3'], df['Weight'], marker='.')
ax1.set_xlabel('Length3 (cm)')
ax1.set_ylabel('Weight (g)')
ax1.set_title('Line Plot')

# 2. Stack Plot → Lengths stacked
ax2.stackplot(df.index[:50], df[['Length1','Length2','Length3']].iloc[:50].T)
ax2.set_xlabel('Index')
ax2.set_ylabel('Lengths')
ax2.set_title('Stack Plot')

# 3. Box Plot → Weight spread
ax3.boxplot(df['Weight'])
ax3.set_xlabel('Fish')
ax3.set_ylabel('Weight (g)')
ax3.set_title('Box Plot')

# 4. Histogram → Weight distribution
ax4.hist(df['Weight'], bins=10, color='skyblue', edgecolor='black')
ax4.set_xlabel('Weight bins')
ax4.set_ylabel('Frequency')
ax4.set_title('Histogram')

# 5. Bar Graph → Avg weight per species
avg_weight = df.groupby('Species')['Weight'].mean()
bars = ax5.bar(avg_weight.index, avg_weight.values)
ax5.set_xlabel('Species')
ax5.set_ylabel('Avg Weight (g)')
ax5.set_title('Bar Graph')
ax5.bar_label(bars, padding=2)
ax5.tick_params(axis='x', rotation=45)

# 6. Violin Plot → Height distribution
ax6.violinplot(df['Height'])
ax6.set_xlabel('Fish')
ax6.set_ylabel('Height (cm)')
ax6.set_title('Violin Plot')

# 7. Horizontal Bar → Species counts
species_counts = df['Species'].value_counts()
bars = ax7.barh(species_counts.index, species_counts.values)
ax7.set_xlabel('Count')
ax7.set_ylabel('Species')
ax7.set_title('Horizontal Bar')
ax7.bar_label(bars, padding=1)

# 8. Scatter Plot → Length vs Width
ax8.scatter(df['Length3'], df['Width'], s=25, c='red')
ax8.set_xlabel('Length3 (cm)')
ax8.set_ylabel('Width (cm)')
ax8.set_title('Scatter Plot')

# 9. Area Plot → Lengths over index
ax9.stackplot(df.index[:50], df[['Length1','Length2','Length3']].iloc[:50].T)
ax9.set_xlabel('Index')
ax9.set_ylabel('Lengths')
ax9.set_title('Area Plot')

plt.suptitle('Different Types of Matplotlib Graphs on Fish Dataset')
plt.subplots_adjust(hspace=0.8, wspace=0.5)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Load and convert image to array
img = Image.open('/content/witcher.webp')
img_arr = np.array(img)

# Top-left: cyan (remove red)
img_arr[47:71, 160:190, 0] = 0

# Top-right: red only (remove green + blue)
img_arr[47:71, 190:215, 1] = 0
img_arr[47:71, 190:215, 2] = 0

# Bottom-left: blue only (remove red + green)
img_arr[71:95, 160:190, 0] = 0
img_arr[71:95, 160:190, 1] = 0

# Bottom-right: green only (remove red + blue)
img_arr[71:95, 190:215, 0] = 0
img_arr[71:95, 190:215, 2] = 0

# Extra step: if any quadrant appears yellow (red+green),
# force it to green by zeroing out red channel
# Example: same bottom-right region, ensure red removed
img_arr[71:95, 190:215, 0] = 0

# Show result
plt.imshow(img_arr)
plt.show()
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Load and convert image to array
img = Image.open('/content/TIGER IMAGE ST.jpg')
img_arr = np.array(img)

# Get dimensions
h, w, _ = img_arr.shape
half_h, half_w = h // 2, w // 2

# Top-left quadrant → cyan (remove red)
img_arr[0:half_h, 0:half_w, 0] = 0

# Top-right quadrant → red (remove green + blue)
img_arr[0:half_h, half_w:w, 1] = 0
img_arr[0:half_h, half_w:w, 2] = 0

# Bottom-left quadrant → blue (remove red + green)
img_arr[half_h:h, 0:half_w, 0] = 0
img_arr[half_h:h, 0:half_w, 1] = 0

# Bottom-right quadrant → green (remove red + blue)
img_arr[half_h:h, half_w:w, 0] = 0
img_arr[half_h:h, half_w:w, 2] = 0

# Show result with axis measures
plt.imshow(img_arr)

plt.grid(False)   # keep grid off but axes visible
plt.show()

```
